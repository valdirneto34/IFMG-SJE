```python
import gurobipy as gp
from gurobipy import GRB
import numpy as np
print(f"Gurobi: {gp.gurobi.version()}")

def otimizar(modelo):
  modelo.optimize()
  if modelo.Status == GRB.INF_OR_UNBD:
    modelo.Params.DualReductions = 0
    modelo.reset()
    modelo.optimize()
  return modelo
```

# Atividade 1 - Produção de produtos de limpeza

```python
'''
max Z = 300x1 + 500x2
sujeito a:
           2x1 + 4x2 <= 40;
           3x1 + 2x2 <= 30;
           x1, x2 >= 0
'''

modelo = gp.Model("modelo")
modelo.Params.OutputFlag = 0
modelo.Params.Method = 1

xA = modelo.addVar(lb=0, name="A")
xB = modelo.addVar(lb=0, name="B")

modelo.setObjective(300*xA + 500*xB, GRB.MAXIMIZE)

modelo.addConstr(2*xA + 4*xB <= 40, name="mistura")
modelo.addConstr(3*xA + 2*xB <= 30, name="materia-prima")

modelo.optimize()

if modelo.Status == GRB.OPTIMAL:
    print(f"\nBásico: {xA.X:.0f} lotes")
    print(f"Premium: {xB.X:.0f} lotes")
    print(f"Margem máxima: R$ {modelo.ObjVal:.2f}/dia")
else:
    print(f"Modelo sem solução ótima. Status: {modelo.Status}")

uso_mistura = 2*xA.X + 4*xB.X
uso_materia = 3*xA.X + 2*xB.X

print(f"\nMistura utilizada: {uso_mistura} de 40")
print(f"Matéria-prima utilizada: {uso_materia} de 30")
```

# Atividade 2 - Composição de uma refeição

```python
'''
min Z = 8x1 + 5x2
sujeito a:
           6x1 + 2x2 >= 30;
           x1 + 4x2 >= 20;
           x1 + x2 = 20;
           x1, x2 >= 0
'''

modelo2 = gp.Model("modelo2")
modelo2.Params.OutputFlag = 0
modelo2.Params.Method = 1

xA = modelo2.addVar(lb=0, name="A")
xB = modelo2.addVar(lb=0, name="B")

modelo2.setObjective(8*xA + 5*xB, GRB.MINIMIZE)

modelo2.addConstr(6*xA + 2*xB >= 30, name="proteina")
modelo2.addConstr(xA + 4*xB >= 20, name="fibras")
modelo2.addConstr(xA + xB == 10, name="quantidade")

modelo2.optimize()

if modelo2.Status == GRB.OPTIMAL:
    print(f"\nAlimento A: {xA.X:.0f} kg")
    print(f"Alimento B: {xB.X:.0f} kg")
    print(f"Custo mínimo: R$ {modelo2.ObjVal:.2f}/dia")
else:
    print(f"Modelo sem solução ótima. Status: {modelo2.Status}")

uso_proteina = 6*xA.X + 2*xB.X
uso_fibras = xA.X + 4*xB.X
quantidade = xA.X + xB.X

print(f"\nProteína utilizada: {uso_proteina} de no mínimo 30")
print(f"Fibra utilizada: {uso_fibras} de no mínimo 20")
print(f"Quantidade utilizada: {quantidade} de exatamente 10")
```

# Atividade 3 - Distribuição de mercadorias

```python
'''
min Z = 4x1 + 6x2 + 8x3 + 5x4 + 3x5 + 7x6
sujeito a:
           x1 + x2 + x3 <= 30;
           x4 + x5 + x6 <= 50;
           x1 + x4 = 30;
           x2 + x5 = 40;
           x3 + x6 = 50;
           x1, x2, x3, x4, x5, x6 >= 0
'''

modelo3 = gp.Model("modelo3")
modelo3.Params.OutputFlag = 0
modelo3.Params.Method = 1

xA = modelo3.addVar(lb=0, name="A")
xB = modelo3.addVar(lb=0, name="B")
xC = modelo3.addVar(lb=0, name="C")

xD = modelo3.addVar(lb=0, name="D")
xE = modelo3.addVar(lb=0, name="E")
xF = modelo3.addVar(lb=0, name="F")

modelo3.setObjective(4*xA + 6*xB + 8*xC + 5*xD + 3*xE + 7*xF, GRB.MINIMIZE)

modelo3.addConstr(xA + xB + xC <= 70, name="oferta_centro1")
modelo3.addConstr(xD + xE + xF <= 50, name="oferta_centro2")

modelo3.addConstr(xA + xD == 30, name="demanda_loja1")
modelo3.addConstr(xB + xE == 40, name="demanda_loja2")
modelo3.addConstr(xC + xF == 50, name="demanda_loja3")

modelo3.optimize()

if modelo3.Status == GRB.OPTIMAL:
    print(f"Centro 1 para Loja 1: {xA.X:.0f} unidades")
    print(f"Centro 1 para Loja 2: {xB.X:.0f} unidades")
    print(f"Centro 1 para Loja 3: {xC.X:.0f} unidades")
    print(f"Centro 2 para Loja 1: {xD.X:.0f} unidades")
    print(f"Centro 2 para Loja 2: {xE.X:.0f} unidades")
    print(f"Centro 2 para Loja 3: {xF.X:.0f} unidades")
    print(f"\nCusto mínimo total: R$ {modelo3.ObjVal:.2f}")
else:
    print(f"Modelo sem solução ótima. Status: {modelo3.Status}")

oferta_c1 = xA.X + xB.X + xC.X
oferta_c2 = xD.X + xE.X + xF.X

demanda_l1 = xA.X + xD.X
demanda_l2 = xB.X + xE.X
demanda_l3 = xC.X + xF.X

print(f"Oferta Centro 1 utilizada: {oferta_c1} de no máximo 70")
print(f"Oferta Centro 2 utilizada: {oferta_c2} de no máximo 50")
print(f"Demanda Loja 1 atendida: {demanda_l1} de exatamente 30")
print(f"Demanda Loja 2 atendida: {demanda_l2} de exatamente 40")
print(f"Demanda Loja 3 atendida: {demanda_l3} de exatamente 50")
```
